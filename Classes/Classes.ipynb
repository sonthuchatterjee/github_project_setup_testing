```python
class Person:
    def __init__(name,name1):
        name.name1 = name1    # bind data to this specific instance

    def greet(self1):
        print(f"Hello, I'm {self1.name1}")

p1 = Person("Alice")
p2 = Person("Bob")

p1.greet()   # uses p1.name
p2.greet()   # uses p2.name
```

```python
# Encapsulation

class Bank:

    def __init__(self,name):
        self.name = name
        self.__balance = 0

    def deposit(self,amount):
        self.__balance+=amount
        

    def withdrawal(self,amount):
        self.__balance-=amount

    def get_balance(self):
        return self.__balance
    

a = Bank('sonthu')
print(a)
a.get_balance()
a.deposit(300)
a.get_balance()
a.withdrawal(200)
a.get_balance()
a.name
#print(a.__balance)


```

```python
#Abstraction

class paymentprocessor:

    def processpayment(self,amount):

        self.__authorization()
        self.__validation()
        self.__sendtobank()
        #self.amount = amount

        print(f"The payment: {amount} has been processed")

    def __authorization(self):
        print(f"Authorization is completed")

    def __validation(self):
        print(f"Validation is completed")

    def __sendtobank(self):
        print(f"Sent to Bank")

b = paymentprocessor()
b.processpayment(100)
#b.amount


```

```python
#Polymorphism Use the same method to act differently

class circle:

    def area(self,radius):
        return 3.17 * (radius**2)

class square:

    def area(self,sidelength):
        return sidelength**2

a=circle()
b=square()

print(a.area(3))
print(b.area(4))

objectlist = [a,b]

for x in objectlist:
    print(x.area(3)) # Polymorphism in action

```

```python
class num1:

    def __init__(self,number):
        self.number1 = number

    def __add__(self,other):
        return self.number1 + other
    
class num2:

    def __init__(self,number):
        self.nummber1 = number

    def __add__(self,other):
        return self.nummber1 + other
    
    def __radd__(self,other):
        return self.nummber1 + other
    
    def __mul__(self,other):
        return self.nummber1 * other
    
    def __truediv__(self,other):
        return self.nummber1 * other
    
    def __floordiv__(self,other):
        return self.nummber1 * other
    
    def __mod__(self,other):
        return self.nummber1 * other

print(num1(1) + num2(2))
```

```python
class person:

    def __init__(self,name,surname,age):

        self.name = name
        self.surname = surname
        self.age = age

    def __repr__(self):
        return (f"person with name: {self.name} and surname: {self.surname} and age: {self.age}")


    def __lt__(self,other):
        return self.age < other.age
    
data = [ {'name':"Pippi", 'surname':'Longstocking', 'age':11}
        , {'name':"Kalle", 'surname':'Blomkvist', 'age':10}
        , {'name':'Karlsson', 'surname': 'on-the-Roof', 'age':12} ]

transformedlist = [person(**row) for row in data]

sorted(transformedlist)

```

```python
class Person:

    def __init__(self,*fishes):
        self.fishes = list(fishes)
    
    def __len__(self):
        return len(self.fishes)
    
    def __getitem__(self,i):
        return self.fishes[i]


print(len(Person('fishes','fish','fishhhh')))

print(Person('fishes','fish','fishhhh')[2])

a= Person()
print(a.__class__)
```
