```python
from fpdf import FPDF

# Create a PDF object with standard A4 size, portrait orientation
pdf = FPDF(format='A4', unit='mm')

# Add a page
pdf.add_page()

# Set font for the title (e.g., Helvetica bold, 16pt)
pdf.set_font("Helvetica", "B", 16)

# Add a cell for the title (width=0 means the rest of the line, height=10, text, border=0, line break=1, align='C')
pdf.cell(0, 10, "Quarterly Sales Report", align="C", ln=True)

# Add a line break
pdf.ln(10)

# Set font for body text (e.g., Helvetica, 12pt)
pdf.set_font("Helvetica", size=12)

# Add some body text
pdf.multi_cell(0, 10, "This report summarizes sales figures for the first quarter.", align="J")
pdf.ln(5)
pdf.multi_cell(0, 10, "Key metrics are detailed below:", align="J")

# You can add more complex elements like tables here using loops and cell methods

# Save the PDF to a file
pdf.output("python_report.pdf")
print("PDF report 'python_report.pdf' generated successfully!")
```
