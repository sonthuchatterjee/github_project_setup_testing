```python
from functools import lru_cache
import requests
import pandas as pd
from csv import DictReader,DictWriter


header = {
    'User-Agent': 'Mozilla/5.0 (Windows NT 10.0; Win64; x64)',
    'Accept': 'application/json',
    'Accept-Language': 'en-US,en;q=0.5',
    'Accept-Encoding': 'gzip, deflate, br',
    'Referer': 'https://example.com',
    'Connection': 'keep-alive'
}
base_url="https://nominatim.openstreetmap.org/search?"

@lru_cache(maxsize=2000)
def nominatim_geocode(baseurl,address,format = 'json', limit=10,headers=header,**kwargs):
    '''thin wrapper around nominatim API'''
    params={'q':address,'format': format,'limit':limit,**kwargs}
    result=requests.get(baseurl,headers=header,params=params)
    return result.json()

def read_csv(filepath):
    with open(filepath,'r') as f:
        #data = list(DictReader(f))
        #data = list(x for x in DictReader(f))
        data = [{i:j  for i,j in x.items() if i != ''} for x in DictReader(f)]
    return data

#print(data)
#print(read_csv(path))

def write_csv(data,outputpath):
    with open(outputpath,'w',newline='') as f:
        writer = DictWriter(f,fieldnames=data[0].keys())

        writer.writeheader()

        #for x in data:
        writer.writerows(data)

# Main  Execution starts here#
response = read_csv('cities.csv')
combineddatadf = pd.DataFrame()
for row in tqdm(response):
    responsedata = nominatim_geocode(base_url,address=row['Name of city'])
    df=pd.json_normalize(responsedata)
    combineddatadf=pd.concat([df,combineddatadf],ignore_index=True)
print(combineddatadf)

combineddatadf.to_csv('geocodedetails_consolidated.csv')
```
