```python
## Customized Iterator
class counter:
    def __init__(self,n):
        self.n = n
        self.counterr = 0
    
    def __iter__(self):
        return self
    
    def __next__(self):
        if self.counterr < self.n:
            value = self.counterr
            self.counterr+=1
            return value
        else:
            raise StopIteration
        
for j in counter(5):
    print(j)
```

```python
class MyRange:
    def __init__(self, start, end):
        self.start = start
        self.end = end
    
    def __iter__(self):
        return self
    
    def __next__(self):
        if self.start < self.end:
            self.start += 1
            return self.start - 1
        raise StopIteration

r = MyRange(1, 4)
for i in r:
    print(i, end=" ")
```

```python
#Generator 

def factors(n):
    for i in range(1,n+1):
        if n % i == 0:
            yield i

a = factors(20)
print(a)
print(list(a))
# print(next(a))
# print(next(a))
# print(next(a))
# print(next(a))
# print(next(a))
# print(next(a))
```
