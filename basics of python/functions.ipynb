```python
def add(x,y):
    return x + y

add(3,y=2)
```

```python
def add(x,y):
    return x + y

add(y=3,2)
```

```python
def add(*args,x,y,):
    return x+y,*args

add(3,4,x=1,y=2)
```

```python
def adding_info(*args,x=0,y=1,):
    if args:
        print(f"{args} are present")
    else:
        print(f"adding the {x} and {y}")

def add(*args,x,y,):
    adding_info(*args,x=2,y=3,)
    return x + y

print(add(3,4,x=1,y=2))
```

```python
#recursive function

def factorial(n):
    if n ==1:
        return n

    print(n*factorial(n-1))
    return n * factorial(n-1)

factorial(3)
```
