```python
def div(x,y):
    try:
        return x/y
    except  ZeroDivisionError as e:
        print(f"you cannot divide a number by zero")
    except Exception as e:
        print(f"The error is due to :{e}")

div(6,0)

div(6,'rrt')
    
```

```python
def div1(x,y):
    try:
        return x/y
    except  ZeroDivisionError as e:
        print(f"you cannot divide a number by zero")
    except Exception as e:
        raise Exception(f"The error is due to :{e}")

div1(6,'rrt')
```

```python
def div2(x,y):
    try:
        return x/y
    except  ZeroDivisionError as e:
        print(f"you cannot divide a number by zero")
    except Exception as e:
        raise ValueError(f"The error is due to :{e}")

div2(6,'rrt')
```
