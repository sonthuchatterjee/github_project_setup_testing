```python
from itertools import cycle
colors = cycle(('red','green','blue'))
nolist = [1,2,3,4,5]

for x,y in zip(nolist,colors):
    print(x,y)
```

```python
from itertools import chain

generator = range(3)
string = "hello"
lis = [1,2,'3a']

for el in chain(generator,string,lis):
    print(el)
```

```python
from itertools import product

s1 = {1,2,3,'4'}
s2 = {'hello',"how"}

for el in product(s1,s2):
    print(el)
```
